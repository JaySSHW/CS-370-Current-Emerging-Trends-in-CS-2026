```python
NB_EPOCH = 20
BATCH_SIZE = 128
VERBOSE = 1
NB_CLASSES = 10 # number of outputs = number of digits
OPTIMIZER = SGD() # SGD optimizer, explained later in this chapter
N_HIDDEN = 128
VALIDATION_SPLIT = 0.2 # how much TRAIN is reserved for VALIDATION
```

---
### Spliting Data Into Train and Test Sets

```python
# data: shuffled and split between train and test sets
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

---
### Shaping

```python
# X_train is 60000 rows of 28x28 values --> reshaped in 60000 x 784

RESHAPED = 784

X_train = X_train.reshape(60000, RESHAPED)
X_test = X_test.reshape(10000, RESHAPED)
X_train = X_train.astype('float32')
X_test = X_test.astype('float32')
```

```python
# normalize
X_train /= 255
X_test /= 255

print(X_train.shape[0], 'train samples')
print(X_test.shape[0], 'test samples')
```

```python
# convert class vectors to binary class matrices

# Y_train = np_utils.to_categorical(y_train, NB_CLASSES) # From the book, code replaced below
Y_train = to_categorical(y_train, NB_CLASSES) # Updated for current version

#Y_test = np_utils.to_categorical(y_test, NB_CLASSES) # From the book, code replaced below
Y_test = to_categorical(y_test, NB_CLASSES) # Updated for current version
```

```python
# Create a neural network with two hidden layers
model = Sequential()

# Input layer
model.add(Input(shape=(RESHAPED,)))

# First hidden layer
model.add(Dense(N_HIDDEN))
model.add(Activation('relu'))

# Second hidden layer
model.add(Dense(N_HIDDEN))
model.add(Activation('relu'))

# Ouput layer
model.add(Dense(NB_CLASSES))
model.add(Activation('softmax'))

# Display the model structure
model.summary()

# Complie the model
model.compile(
    loss='categorical_crossentropy',
    optimizer=OPTIMIZER,
    metrics=['accuracy']
)

# Train the model
history = model.fit(
    X_train,
    Y_train,
    batch_size=BATCH_SIZE,
    epochs=NB_EPOCH,
    verbose=VERBOSE,
    validation_split=VALIDATION_SPLIT
)

# Evaluate the model using the test data
score = model.evaluate(X_test, Y_test, verbose=VERBOSE)

print("test Score:", score[0])
print("test Accuracy:", score[1])
print("Training Accuracy:", history.history['accuracy'][-1])
print("Validation Accuracy:", history.history['val_accuracy'][-1])
```

### Analysis of Accuracy Results

For these experiments, I changed the number of neurons in each hidden layer while keeping the other network and training parameters the same. The baseline model used 128 hidden neurons in each layer and achieved a training accuracy of 94.53%, a validation accuracy of 94.73%, and a test accuracy of 94.25%. The similar training, validation, and test accuracy rates indicate that the model was able to perform consistently on both the data used during training and data it had not directly trained on.

In Experiment 1, I reduced the number of hidden neurons from 128 to 32. The training accuracy decreased to 93.42%, validation accuracy decreased to 93.57%, and the test accuracy decreased to 93.49%. Compared with the baseline, reducing the number of neurons lowered accuracy across all three data sets. This suggests that the smaller network had less capacity to learn the patterns needed to distinguish the handwritten digits.

In Experiment 2, I increased the number of hidden neurons from 128 to 512. The training accuracy increased to 95.38%, validation accuracy increased to 95.32%, and the test accuracy increased to 95.06%. The 512-neuron model therefore achieved the highest accuracy across all three data sets. The test accuracy improved by approximately 1.57 percentage points compared with the 32-neuron model and by approximately 0.81 percentage points compared with the baseline model. Increasing the number of hidden neurons gave the network greater capacity to learn useful patterns in the MNIST images. 

Overall, the results demonstrate a clear relationship between the number of hidden neurons and accuracy in these experiments. The 32-neuron model produced the lowest accuracy, the 128-neuron baseline performed better, and the 512-neuron model produced the highest accuracy. Training, validation, and test accuracy also remained relatively close for all three configurations, so these results do not show a large gap that would suggest substantial overfitting. However, increasing the number of hidden neurons also increases the number of trainable parameters and computational requirements. Therefore, although increasing network capacity improved accuracy in these experiments, a larger neural network also requires additional computational resourses and may eventually produce diminishing improvements.
